```python
from pathlib import Path
import nbformat


project_path = Path(r"D:\final_project_depi")

notebooks_path = project_path / "notebooks"


notebooks_path.mkdir(parents=True, exist_ok=True)


notebook_names = [
    "01_EDA.ipynb",
    "02_Data_Cleaning.ipynb",
    "03_Feature_Engineering.ipynb",
    "04_Preprocessing.ipynb",
    "05_Modeling.ipynb"
]


for name in notebook_names:
    file_path = notebooks_path / name

    if not file_path.exists():
        nb = nbformat.v4.new_notebook()
        nb.cells.append(
            nbformat.v4.new_markdown_cell(f"# {name.replace('.ipynb', '')}")
        )
        nbformat.write(nb, file_path)
        print(f"Created: {name}")
    else:
        print(f"Already exists: {name}")

print("\nAll notebooks are ready!")
```
